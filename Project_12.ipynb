Understanding Convolution and Pooling

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load image
img = cv2.imread('/content/image.jpg', 0)

# Convolution filter
kernel = np.array([[-1, -1, -1],
                   [-1,  8, -1],
                   [-1, -1, -1]])

conv = cv2.filter2D(img, -1, kernel)
```

```python
# Max pooling
h, w = img.shape
h, w = h // 2, w // 2
pool = img[:h*2, :w*2].reshape(h, 2, w, 2).max(axis=(1, 3))

# Display results
plt.figure(figsize=(10, 5))
```

```python
plt.subplot(1, 3, 1)
plt.imshow(img, cmap='gray')
plt.title('Original Image')
plt.axis('off')
```

```python
plt.subplot(1, 3, 2)
plt.imshow(conv, cmap='gray')
plt.title('Convolution')
plt.axis('off')
```

```python
plt.subplot(1, 3, 3)
plt.imshow(pool, cmap='gray')
plt.title('Max Pooling')
plt.axis('off')
```

```python
plt.tight_layout()
plt.show()
```
