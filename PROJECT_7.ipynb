```python
import numpy as np

actual = np.array([1, 0, 1, 1, 0])
predicted = np.array([0.9, 0.2, 0.8, 0.7, 0.1])

# Mean Squared Error
mse = np.mean((actual - predicted) ** 2)

# Cross-Entropy Loss
epsilon = 1e-15
predicted = np.clip(predicted, epsilon, 1 - epsilon)

cross_entropy = -np.mean(
    actual * np.log(predicted) +
    (1 - actual) * np.log(1 - predicted)
)

print("Mean Squared Error:", mse)
print("Cross-Entropy Loss:", cross_entropy)
```

Comparison

MSE	Measures the squared difference between actual and predicted values and 	Regression problems

Cross-Entropy Loss	Measures how well predicted probabilities match the actual class labels and	Classification problems


Explanation:

Mean Squared Error (MSE) calculates the average squared difference between the actual values and predicted values. It is commonly used for regression, where the output is a continuous numerical value.

Cross-Entropy Loss is commonly used for classification. It works with predicted probabilities and gives a larger penalty when the model is confidently wrong.


Conclusion:

The two loss functions measure prediction errors differently. MSE is generally suitable for regression, while Cross-Entropy Loss is generally preferred for classification tasks because it directly evaluates the predicted probabilities of the classes.
