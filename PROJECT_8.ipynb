Gradient Descent Implementation

```python
import numpy as np
import matplotlib.pyplot as plt

# Simple dataset
X = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 6, 8, 10])

# Initialize parameters
m = 0
b = 0

# Learning rate
learning_rate = 0.01

# Number of iterations
iterations = 1000

# Store loss values
loss_history = []

# Gradient Descent
n = len(X)

for i in range(iterations):

    # Predictions
    y_pred = m * X + b

    # Calculate error
    error = y_pred - y

    # Calculate gradients
    dm = (2/n) * np.sum(X * error)
    db = (2/n) * np.sum(error)

    # Update parameters
    m = m - learning_rate * dm
    b = b - learning_rate * db

    # Calculate loss
    loss = np.mean(error ** 2)
    loss_history.append(loss)

# Display final parameters
print("Final slope (m):", m)
print("Final intercept (b):", b)
print("Final loss:", loss_history[-1])

# Plot loss reduction
plt.plot(loss_history)
plt.xlabel("Iterations")
plt.ylabel("Loss")
plt.title("Loss Reduction using Gradient Descent")
plt.show()
```

Analysis

Gradient Descent is an optimization algorithm used to reduce the error of a model. Initially, the parameters m and b are set to zero. In every iteration, the model calculates predictions, measures the error, calculates the gradients, and updates the parameters. As the iterations increase, the loss decreases and the model's predictions become closer to the actual values.

Result:

The experiment demonstrates that Gradient Descent successfully updates the model parameters and reduces prediction error over multiple iterations.
